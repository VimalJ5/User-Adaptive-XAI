```python
from datasets import load_dataset
import pandas as pd

# Load 500 samples
ds = load_dataset("sentence-transformers/eli5", split="train[:500]")

# Convert to pandas
df = pd.DataFrame(ds)

# Keep only relevant columns
df = df[["question", "answer"]]

# Save to CSV
df.to_csv("eli5_500.csv", index=False)

print("Saved 500 samples to eli5_500.csv")
```
