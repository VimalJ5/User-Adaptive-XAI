```python
import json
from pathlib import Path

from textstat import textstat


results_path = Path("results_expert.json")
with results_path.open("r", encoding="utf-8") as file:
    results = json.load(file)

explanations = [
    item["explanation"]
    for item in results
    if isinstance(item, dict) and isinstance(item.get("explanation"), str) and item["explanation"].strip()
]

fre_scores = [textstat.flesch_reading_ease(text) for text in explanations]
fkgl_scores = [textstat.flesch_kincaid_grade(text) for text in explanations]

average_fre = sum(fre_scores) / len(fre_scores) if fre_scores else float("nan")
average_fkgl = sum(fkgl_scores) / len(fkgl_scores) if fkgl_scores else float("nan")

print(f"Average FRE: {average_fre:.2f}")
print(f"Average FKGL: {average_fkgl:.2f}")
```
